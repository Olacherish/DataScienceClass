## **STEP TWO: SPLITTING THE DATASETS**

```
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer

# Load the dataset and define x (features) and y (target)
data = load_breast_cancer(as_frame=True)
x = data.data
y = data.target

# Split the x, y data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.30, random_state=42)
```

## **STEP THREE: TRAIN THE MODEL**

```
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Instantiate the Logistic Regression model
# (Assuming logistic_regression_model was defined in a previous cell and is available)
logistic_regression_model = LogisticRegression(max_iter=20000, random_state=42)

# Train the model
logistic_regression_model.fit(x_train, y_train)

# Make predictions on the test set
y_pred = logistic_regression_model.predict(x_test)

# Evaluate the model's accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}")
```

### Feature Importance from Model Coefficients

```
import pandas as pd

# Get the coefficients from the trained model
coefficients = logistic_regression_model.coef_[0]

# Get the feature names from the training data
feature_names = x_train.columns

# Create a DataFrame to display feature importance
feature_importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': coefficients
})

# Sort by the absolute value of the coefficients to see the most impactful features
feature_importance_df['Absolute_Coefficient'] = feature_importance_df['Coefficient'].abs()
feature_importance_df = feature_importance_df.sort_values(by='Absolute_Coefficient', ascending=False)

display(feature_importance_df.drop(columns=['Absolute_Coefficient']))
```
