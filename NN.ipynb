```python
# MLP STEPS

1)Forward pass (forward propapagation)
from front to back i.e from input to output and
the final result (end result) is PREDICTION
# Randomization of weight & bias

2) Loss calculation (cross entropy,log loss)
predictions are compared to the true labels
using LOSS FUNCTION.
# Loss function changes the values of weights and biases

The loss guide the optimization to know how far is
the PREDICTION from the TARGET

# Note when we have above 5 hidden layers
we call it-> DEEP NEURAL NETWORK (Deep Learning)

3) backpropagation
moving from (output_1) or prediction to the first hidden layer

```

```python
# We have 2 stages/phases  of AI Model
- Training stage/phase  (model is undergoing training)
- Inference  "        (after training model is used to make prediction)
```

- **Loss Functions**: Customize for specific tasks (e.g., focal loss for imbalanced data).
- **Optimizers**: Explore SGD with momentum, RMSprop, or A <a id="section8"><a/>Adam for better convergence.
- **Deep Architectures**: Extend MLPs to deeper networks or specialized architectures like CNNs or RNNs.
