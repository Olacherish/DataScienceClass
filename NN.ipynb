## **BACKPROPAGATION**

```python
# The goal of backpropagation is to MINIMIZE the LOSS Function/ Error

Model Hallucination is an error (rubbish)

# Adjusting weights and biases occurs during backpropagation and optimization.
# Calculating gradients of the loss function is also part of backpropagation.

# Forward pass ends in -> PREDICTION
E.g image_1: car images
    image_2: car images
```

```python
1. Forward pass
the Math -> is Linear Algebra # calculus # Z = W1X1 + W2X2 + ...... + WnXn + b

# Compute the output each layer
# Calculate loss

2. Backward pass
# Compute the Gradient of the loss with respect to Output
# Propagate Gradients backward thru Layers using the chain rule
  - for each layer
  - Update weights
  - Update biases
3. Weight Update:
  - use Gradient descent
  - similarity update Biases

4. Iterate: repeat forward and backward passes over
multiple Epochs until convergent
```

## **KEY COMPONENTS**

```python
1. Loss function
  - Regression: Mean Squared Error
  - Classification: Cross Entropy Loss

2. Gradient Descent variants:algorithm use in Backpropagation to minimize loss fxn
  - Batch Gradient Descents: Uses the Entire datasets(Stable but Slow)

  - Stoichastic Gradient Descent: Uses a single/one sample at a time (last but Noisy)
# Stoichastic is the same as Randomization
  - Mini-Batch SDG: Uses small batch (balance stable,speed)

3. Learning Rate
  Control step-sizes
  - Too High -> Divergent
  - Too Low -> Slow convergence

4. Regularization
```
